```sql

CREATE OR REPLACE TABLE fuel_price_dev.gold.agg_prix_moyen_par_departement_jour AS
SELECT 
s.code_departement,
c.nom,
CAST(p.date_maj AS DATE) AS date,
MIN(p.prix) AS prix_min,
MAX(p.prix) AS prix_max,
COUNT(s.station_id) AS nb_stations,
ROUND(AVG(p.prix),2) AS prix_moyen,
current_date() AS date_traitement
FROM fuel_price_dev.silver.fait_prix p
LEFT JOIN fuel_price_dev.silver.dim_station s ON s.station_id = p.station_id
LEFT JOIN fuel_price_dev.silver.dim_carburant c ON p.carburant_id = c.id
GROUP BY s.code_departement, c.nom,CAST(p.date_maj AS DATE)


```
