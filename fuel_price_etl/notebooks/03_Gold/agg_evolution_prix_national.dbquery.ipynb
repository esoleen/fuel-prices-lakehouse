```sql
CREATE OR REPLACE TABLE fuel_price_dev.gold.agg_evolution_prix_national AS
SELECT 
CAST(p.date_maj AS DATE) AS date,
c.nom AS carburant,
ROUND(AVG(p.prix),2) AS prix_moyen_national
FROM  fuel_price_dev.silver.fait_prix p 
LEFT JOIN fuel_price_dev.silver.dim_carburant c ON c.id = p.carburant_id 
GROUP BY CAST(p.date_maj AS DATE), c.nom
```
