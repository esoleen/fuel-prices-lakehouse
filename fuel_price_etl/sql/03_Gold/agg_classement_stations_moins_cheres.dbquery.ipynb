```sql
CREATE OR REPLACE TABLE fuel_price_dev.gold.agg_classement_stations_moins_cheres AS 
with cte_classement_carburant AS(
    SELECT 
    d.code_departement,
    g.departement ,
    c.nom AS carburant,
    d.station_id,
    p.prix,
    ROW_NUMBER()OVER(PARTITION BY d.code_departement,c.nom  ORDER BY p.prix ASC) as rang
    FROM  fuel_price_dev.silver.fait_prix p
    LEFT JOIN fuel_price_dev.silver.dim_carburant c ON c.id = p.carburant_id
    LEFT JOIN fuel_price_dev.silver.dim_station d ON d.station_id = p.station_id
    LEFT JOIN fuel_price_dev.silver.dim_geo g ON g.code_departement = d.code_departement
)
SELECT 
    code_departement,
    departement,
    carburant,
    station_id,
    prix,
    rang
FROM cte_classement_carburant
WHERE rang <= 5
ORDER BY code_departement, carburant, rang;
```
