```python

SELECT 
CAST(p.date_maj AS DATE) AS date,
c.nom AS nom_carburant,
MIN(p.prix)OVER() AS prix_moyen_national
 FROM  fuel_price_dev.silver.fait_prix p 
LEFT JOIN fuel_price_dev.silver.dim_carburant c ON s.carburant_id = p.carburant_id 
LEFT JOIN fuel_price_dev.silver.dim_station s ON c.station_id = p.station_id AND s.carburant_id = p.carburant_id 
LEFT JOIN fuel_price_dev.silver.dim_geo g ON s.code_region = g.code_region
```
